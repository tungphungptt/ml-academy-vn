Phân cụm 3 nhóm, báo cáo inertia.

Chạy từ thư mục gốc repository với môi trường trong `requirements-curriculum.txt`. Với PyTorch, cài CPU wheel từ index chính thức như hướng dẫn trong docs/curriculum-practice.md. Output bên dưới được ghi từ lần chạy CPU thực tế; không phải benchmark phần cứng.


```python
import numpy as np
rng = np.random.default_rng(42)
X = np.vstack([rng.normal(c, 0.6, (40, 2)) for c in ([0, 0], [5, 0], [2.5, 5])])
C = X[rng.choice(len(X), 3, replace=False)]
for _ in range(20):
    a = np.argmin(((X[:, None] - C[None]) ** 2).sum(-1), axis=1)
    C = np.array([X[a == k].mean(0) if (a == k).any() else C[k] for k in range(3)])
print(f"Inertia: {((X - C[a]) ** 2).sum():.2f}")
```
