Baseline: Logistic Regression có chuẩn hóa · Chỉ số: Recall, F1

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
import joblib, os, tempfile
X, y = load_breast_cancer(return_X_y=True)
Xa, Xb, ya, yb = train_test_split(X, y, test_size=0.25, random_state=0, stratify=y)
```

```python
pipe = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000))
gs = GridSearchCV(pipe, {'logisticregression__C': [0.01, 0.1, 1, 10]}, cv=5, scoring='recall').fit(Xa, ya)
print('Tham số tốt nhất:', gs.best_params_, '| recall CV:', round(gs.best_score_, 3))
```

```python
from sklearn.metrics import recall_score, f1_score
p = gs.predict(Xb)
print('Test: recall =', round(recall_score(yb, p), 3), '| F1 =', round(f1_score(yb, p), 3))
f = os.path.join(tempfile.mkdtemp(), 'model.joblib'); joblib.dump(gs.best_estimator_, f)
print('Nạp lại cho cùng dự đoán:', bool((joblib.load(f).predict(Xb) == p).all()))
```
