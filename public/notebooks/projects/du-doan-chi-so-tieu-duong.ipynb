1. Tải dữ liệu và chia train/test (random_state cố định).
2. Huấn luyện baseline: dự đoán bằng giá trị trung bình.
3. Huấn luyện LinearRegression và Ridge.
4. So sánh MSE/R² trên tập test và nhận xét.

Baseline: Dự đoán bằng trung bình của tập train · Chỉ số: MSE, R²

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error, r2_score
X, y = load_diabetes(return_X_y=True)
Xa, Xb, ya, yb = train_test_split(X, y, test_size=0.25, random_state=0)
print(X.shape, y.shape)
```

```python
models = {'baseline (trung bình)': DummyRegressor(), 'LinearRegression': LinearRegression(), 'Ridge(alpha=1)': Ridge(alpha=1.0)}
for name, m in models.items():
    p = m.fit(Xa, ya).predict(Xb)
    print(f'{name:24s} MSE={mean_squared_error(yb, p):8.1f}  R2={r2_score(yb, p):.3f}')
```
