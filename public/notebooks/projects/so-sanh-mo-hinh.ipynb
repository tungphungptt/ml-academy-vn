# So sánh nhiều mô hình trên cùng một dataset

Notebook khởi đầu của ML Academy VN. **Các kết quả bên dưới là kết quả thật khi chạy bằng Python 3 + scikit-learn lúc phát triển** (số có thể lệch nhẹ giữa các phiên bản thư viện). Hãy chạy lại, chỉnh sửa và hoàn thành các bước:

1. Chia train/test một lần với random_state cố định.
2. Dựng Pipeline chuẩn hóa + mô hình cho từng thuật toán.
3. Chạy 5-fold cross-validation trên train, ghi trung bình và độ lệch chuẩn.
4. Đánh giá mô hình tốt nhất trên test đúng một lần.
5. Nhận xét: chênh lệch nhỏ có đáng tin không?

Baseline: DummyClassifier(most_frequent) · Chỉ số: Accuracy (CV), F1 macro

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.dummy import DummyClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
X, y = load_digits(return_X_y=True)
Xa, Xb, ya, yb = train_test_split(X, y, test_size=0.25, random_state=0, stratify=y)
```

```python
models = {'baseline': DummyClassifier(strategy='most_frequent'), 'KNN': KNeighborsClassifier(5), 'LogReg': LogisticRegression(max_iter=2000), 'SVM-rbf': SVC(), 'RandomForest': RandomForestClassifier(n_estimators=200, random_state=0)}
res = {}
for name, m in models.items():
    s = cross_val_score(make_pipeline(StandardScaler(), m), Xa, ya, cv=5, scoring='accuracy')
    res[name] = s.mean(); print(f'{name:12s} acc(CV)={s.mean():.3f} ± {s.std():.3f}')
```

```python
best = max(res, key=res.get)
m = make_pipeline(StandardScaler(), models[best]).fit(Xa, ya)
print('Mô hình tốt nhất theo CV:', best, '| accuracy trên test (dùng 1 lần):', round(m.score(Xb, yb), 3))
```
