4. Thử MLP (scikit-learn hoặc PyTorch) và so sánh.

Baseline: KNN với k=5 · Chỉ số: Accuracy, confusion matrix

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
X, y = load_digits(return_X_y=True)
Xa, Xb, ya, yb = train_test_split(X / 16.0, y, test_size=0.25, random_state=0, stratify=y)
```

```python
models = {'KNN k=5': KNeighborsClassifier(5), 'SVM-rbf': SVC(), 'MLP': MLPClassifier(hidden_layer_sizes=(64,), max_iter=500, random_state=0)}
for name, m in models.items():
    print(f'{name:10s} accuracy={accuracy_score(yb, m.fit(Xa, ya).predict(Xb)):.3f}')
```

```python
import numpy as np
cm = confusion_matrix(yb, SVC().fit(Xa, ya).predict(Xb))
np.fill_diagonal(cm, 0)
i, j = np.unravel_index(cm.argmax(), cm.shape)
print('Cặp bị nhầm nhiều nhất (thật -> đoán):', i, '->', j, 'số lần:', cm[i, j])
```
